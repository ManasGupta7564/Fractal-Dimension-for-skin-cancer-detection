```python
import numpy as np
import pandas as pd
from tensorflow.keras.datasets import mnist

# ==========================
# Load MNIST labels
# ==========================
(_, _), (_, y_test) = mnist.load_data()

# ==========================
# CSV files
# ==========================
csv_files = [
    "/home/ubuntu/local/MNIST/features_16245_reordered.csv",  # Fractal (T=30)
    "/home/ubuntu/local/runs/CL200_T10_S7.5_E50_20260621_225246/Class_Sums/features_8820.csv",  # HOG (T=10)
    "/home/ubuntu/local/runs/CL200_T10_S7.5_E50_20260626_220724/Class_Sums/features_20480.csv"  # Threshold (T=10)
]

dfs = [pd.read_csv(f) for f in csv_files]

# ==========================
# Extract class scores
# ==========================
fractal = dfs[0][[f"Class_{i}" for i in range(10)]].values.astype(np.float32)
hog = dfs[1][[f"Class_{i}" for i in range(10)]].values.astype(np.float32)
threshold = dfs[2][[f"Class_{i}" for i in range(10)]].values.astype(np.float32)

# ==========================
# Normalize by Threshold (T)
# ==========================
fractal = fractal / 30.0
hog = hog / 10.0
threshold = threshold / 10.0

# ==========================
# Combine normalized votes
# ==========================
total_class_votes = fractal + hog + threshold

overall_preds = np.argmax(total_class_votes, axis=1)

accuracy = np.mean(overall_preds == y_test)

print(f"Overall Accuracy: {accuracy * 100:.2f}%")
```
