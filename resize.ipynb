```python
import os
import cv2
from tqdm import tqdm

# INPUT: original ISIC image folder
input_dir = "/home/ubuntu/Downloads/ISIC2019/ISIC_2019_Training_Input/ISIC_2019_Training_Input"

# OUTPUT: resized image folder
output_dir = "/home/ubuntu/Downloads/ISIC2019/ISIC_2019_Training_Input_resized"

os.makedirs(output_dir, exist_ok=True)

TARGET_WIDTH = 600
TARGET_HEIGHT = 450
TARGET_SIZE = (TARGET_WIDTH, TARGET_HEIGHT)

image_files = [f for f in os.listdir(input_dir) if f.endswith(".jpg")]

print("Total images found:", len(image_files))

resized_count = 0
skipped_count = 0

for img_name in tqdm(image_files):

    input_path = os.path.join(input_dir, img_name)
    output_path = os.path.join(output_dir, img_name)

    img = cv2.imread(input_path)

    if img is None:
        print("Skipping unreadable file:", img_name)
        continue

    h, w = img.shape[:2]

    # Skip resizing if already correct size
    if (w, h) == (TARGET_WIDTH, TARGET_HEIGHT):
        cv2.imwrite(output_path, img)
        skipped_count += 1
        continue

    # Resize image
    resized_img = cv2.resize(img, TARGET_SIZE, interpolation=cv2.INTER_AREA)

    cv2.imwrite(output_path, resized_img)
    resized_count += 1


print("\nResizing completed.")
print("Images resized:", resized_count)
print("Images already correct size (skipped):", skipped_count)
```
