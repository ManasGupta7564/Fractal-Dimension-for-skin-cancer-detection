```python
import numpy as np
from Booleanize_global import *
```

```python
X_train_fd = np.load("X_train_fd.npy")
X_test_fd = np.load("X_test_fd.npy")

y_train = np.load("y_train.npy")
y_test = np.load("y_test.npy")

print("Train shape:", X_train_fd.shape)
print("Test shape:", X_test_fd.shape)
```

```python
all_train_values = X_train_fd.flatten()

print("Total values used for binning:", len(all_train_values))

discretizer = fit_global_bins(all_train_values, n_bins=5)
```

```python
print("\nGlobal Quantile Bin Boundaries:")
bins = discretizer.bin_edges_[0]

for i in range(len(bins) - 1):
    print(f"Bin {i}: {bins[i]:.6f} → {bins[i+1]:.6f}")
```

```python
X_train_bin = booleanize_array(X_train_fd, discretizer)
X_test_bin  = booleanize_array(X_test_fd, discretizer)

print("\nAfter booleanization:")
print("X_train_bin shape:", X_train_bin.shape)
print("X_test_bin shape:", X_test_bin.shape)
```

X_train_bin = np.concatenate([X_train_bin, 1 - X_train_bin], axis=1)
X_test_bin  = np.concatenate([X_test_bin, 1 - X_test_bin], axis=1)

print("\nAfter adding negation:")
print("X_train_bin shape:", X_train_bin.shape)
print("X_test_bin shape:", X_test_bin.shape)

```python
np.save("X_train_bin_cifar.npy", X_train_bin)
np.save("X_test_bin_cifar.npy", X_test_bin)
np.save("y_train_cifar.npy", y_train)
np.save("y_test_cifar.npy", y_test)

print("\nSaved: X_train_bin.npy, X_test_bin.npy, y_train.npy, y_test.npy")
```
