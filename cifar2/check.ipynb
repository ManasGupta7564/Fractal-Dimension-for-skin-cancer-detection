```python
import pandas as pd
import numpy as np
import os

# ==========================
# Fractal class sum CSV
# ==========================
votes = pd.read_csv(
    "/home/ubuntu/local/ShellScripted_TA/runs/CL400_T20_S7.5_E30_20260621_011345/Class_Sums/features_16245.csv"
)

# ==========================
# Folder containing individual fractal CSVs
# ==========================
csv_dir = "/home/ubuntu/local/cifar2/fractal_maps_cifar_resized_M8_S1"

csv_files = [f for f in os.listdir(csv_dir) if f.endswith(".csv")]
test_files = [f for f in csv_files if f.startswith("test_")]

print("Test files :", len(test_files))
print("Vote rows  :", len(votes))

assert len(test_files) == len(votes)

# ==========================
# Reorder class sums
# ==========================
reordered = np.zeros_like(votes.values)

for current_row, filename in enumerate(test_files):

    # Example filename:
    # test_0_523_fd.csv
    # test_1_7812_fd.csv

    original_index = int(filename.split("_")[2])

    reordered[original_index] = votes.values[current_row]

# ==========================
# Save reordered CSV
# ==========================
os.makedirs("/home/ubuntu/local/cifar2", exist_ok=True)

reordered_df = pd.DataFrame(reordered, columns=votes.columns)

save_path = "/home/ubuntu/local/cifar2/features_16245_reordered.csv"

reordered_df.to_csv(save_path, index=False)

print(f"\n✅ Reordered class sums saved to:\n{save_path}")
```
