```python
import os
import cv2
from tensorflow.keras.datasets import mnist

# =========================
# LOAD MNIST
# =========================
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# =========================
# BASE DIRECTORY
# =========================
base_dir = "/home/ubuntu/mnist_images"

# =========================
# CREATE FOLDER STRUCTURE
# =========================
for split in ["train", "test"]:
    for label in range(10):
        path = os.path.join(base_dir, split, str(label))
        os.makedirs(path, exist_ok=True)

print("Folders created!")

# =========================
# SAVE TRAIN IMAGES
# =========================
for idx, (img, label) in enumerate(zip(x_train, y_train)):
    
    # Convert grayscale → RGB (important for your C++ pipeline)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
    
    save_path = os.path.join(base_dir, "train", str(label), f"{idx}.png")
    cv2.imwrite(save_path, img_rgb)

    if (idx + 1) % 5000 == 0:
        print(f"Train: {idx+1}/{len(x_train)} saved")

# =========================
# SAVE TEST IMAGES
# =========================
for idx, (img, label) in enumerate(zip(x_test, y_test)):
    
    img_rgb = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
    
    save_path = os.path.join(base_dir, "test", str(label), f"{idx}.png")
    cv2.imwrite(save_path, img_rgb)

    if (idx + 1) % 2000 == 0:
        print(f"Test: {idx+1}/{len(x_test)} saved")

print("✅ MNIST dataset saved successfully!")
```

```python
import os
import cv2
import ctypes
import numpy as np
from tensorflow.keras.datasets import mnist

# =========================
# LOAD RESIZE LIBRARY
# =========================
lib_resize = ctypes.CDLL("./libresize.so")

lib_resize.resize_image.argtypes = [
    ctypes.POINTER(ctypes.c_double),  # input
    ctypes.c_int,                     # old_h
    ctypes.c_int,                     # old_w
    ctypes.POINTER(ctypes.c_double),  # output
    ctypes.c_int,                     # new_h
    ctypes.c_int                      # new_w
]

lib_resize.resize_image.restype = None

# =========================
# LOAD MNIST
# =========================
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# =========================
# OUTPUT DIRECTORY
# =========================
base_dir = "/home/ubuntu/mnist_images_resized"

# =========================
# CREATE FOLDER STRUCTURE
# =========================
for split in ["train", "test"]:
    for label in range(10):
        os.makedirs(
            os.path.join(base_dir, split, str(label)),
            exist_ok=True
        )

print("Folders created!")

# =========================
# RESIZE FUNCTION
# Uses libresize.so
# Output remains RGB (3 channels)
# =========================
def resize_mnist_lib(img):
    
    # Convert grayscale -> RGB
    img_rgb = np.stack([img] * 3, axis=-1).astype(np.float64)

    old_h, old_w = img_rgb.shape[:2]

    # Output buffer
    out = np.zeros((64, 64, 3), dtype=np.float64)

    # Call C++ resize
    lib_resize.resize_image(
        img_rgb.ravel().ctypes.data_as(ctypes.POINTER(ctypes.c_double)),
        old_h,
        old_w,
        out.ravel().ctypes.data_as(ctypes.POINTER(ctypes.c_double)),
        64,
        64
    )

    return np.clip(out, 0, 255).astype(np.uint8)

# =========================
# SAVE TRAIN IMAGES
# =========================
for idx, (img, label) in enumerate(zip(x_train, y_train)):

    img_resized = resize_mnist_lib(img)

    save_path = os.path.join(
        base_dir,
        "train",
        str(label),
        f"{idx}.png"
    )

    cv2.imwrite(save_path, img_resized)

    if (idx + 1) % 5000 == 0:
        print(f"Train: {idx+1}/{len(x_train)} saved")

# =========================
# SAVE TEST IMAGES
# =========================
for idx, (img, label) in enumerate(zip(x_test, y_test)):

    img_resized = resize_mnist_lib(img)

    save_path = os.path.join(
        base_dir,
        "test",
        str(label),
        f"{idx}.png"
    )

    cv2.imwrite(save_path, img_resized)

    if (idx + 1) % 2000 == 0:
        print(f"Test: {idx+1}/{len(x_test)} saved")

print("✅ MNIST resized to 64×64 using libresize.so and saved successfully!")

# =========================
# OPTIONAL VERIFICATION
# =========================
sample = cv2.imread(
    os.path.join(base_dir, "train", "0", "1.png"),
    cv2.IMREAD_UNCHANGED
)

print("\nVerification:")
print("Shape:", sample.shape)
print("Dtype:", sample.dtype)

if len(sample.shape) == 3:
    print("Channels:", sample.shape[2])
    print("R==G:", np.all(sample[:,:,0] == sample[:,:,1]))
    print("G==B:", np.all(sample[:,:,1] == sample[:,:,2]))
```
