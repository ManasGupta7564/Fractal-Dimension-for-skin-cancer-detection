```python
import pandas as pd
import os

# Your class-sum CSV
votes = pd.read_csv("/home/ubuntu/local/ShellScripted_TA/runs/CL2500_T30_S7.5_E30_20260621_011812/Class_Sums/features_16245.csv")

csv_dir = "fractal_maps_mnist_resized_M8_S1"

csv_files = [f for f in os.listdir(csv_dir) if f.endswith(".csv")]
test_files = [f for f in csv_files if f.startswith("test_")]

# Allocate output
reordered = [None] * len(test_files)

for current_row, filename in enumerate(test_files):
    original_index = int(filename.split("_")[2])   # <-- MNIST index

    reordered[original_index] = votes.iloc[current_row]

reordered_df = pd.DataFrame(reordered)
reordered_df.to_csv("features_16245_reordered.csv", index=False)

print("Done!")
```

```python
import pandas as pd
import numpy as np
import os

votes = pd.read_csv("/home/ubuntu/local/ShellScripted_TA/runs/CL2500_T30_S7.5_E30_20260621_011812/Class_Sums/features_16245.csv")


csv_dir = "fractal_maps_mnist_resized_M8_S1"
test_files = [f for f in os.listdir(csv_dir) if f.startswith("test_") and f.endswith(".csv")]

assert len(votes) == len(test_files)

reordered = np.zeros_like(votes.values)

for current_row, filename in enumerate(test_files):
    original_index = int(filename.split("_")[2])
    reordered[original_index] = votes.values[current_row]

reordered_df = pd.DataFrame(reordered, columns=votes.columns)
reordered_df.to_csv("features_16245_reordered.csv", index=False)
```
