```python
import pandas as pd
import numpy as np
import os

# ==========================
# Fractal class sum CSV
# ==========================
votes = pd.read_csv(
    "/home/ubuntu/local/ShellScripted_TA/runs/CL2500_T30_S7.5_E30_20260621_011124/Class_Sums/features_16245.csv"
)

# ==========================
# Folder containing individual fractal CSVs
# ==========================
csv_dir = "fractal_maps_fmnist_M8_S1"   # <-- change if different

csv_files = [f for f in os.listdir(csv_dir) if f.endswith(".csv")]
test_files = [f for f in csv_files if f.startswith("test_")]

print("Test files:", len(test_files))
print("Vote rows :", len(votes))

assert len(test_files) == len(votes)

# ==========================
# Reorder
# ==========================
reordered = np.zeros_like(votes.values)

for current_row, filename in enumerate(test_files):

    # Filename:
    # test_7_523_fd.csv
    original_index = int(filename.split("_")[2])

    reordered[original_index] = votes.values[current_row]

# ==========================
# Save
# ==========================
reordered_df = pd.DataFrame(reordered, columns=votes.columns)

reordered_df.to_csv(
    "/home/ubuntu/local/fmnist/features_16245_reordered.csv",
    index=False
)

print("Done!")
```
